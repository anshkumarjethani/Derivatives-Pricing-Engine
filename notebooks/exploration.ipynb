```python
from scipy.stats import norm
```

```python
norm.cdf(0)
```

```python
S = 100 
K = 100
r = 0.05
sigma = 0.2
T = 1
```

```python
import numpy as np
```

```python
d1 = (np.log(S/K) + (r + sigma**2 / 2)) * T / (sigma * np.sqrt(T))
d2 = d1 - sigma * np.sqrt(T)

d1, d2
```

```python
N_d1 = norm.cdf(d1)
N_d2 = norm.cdf(d2)

N_d1, N_d2
```

```python
call_price = S * N_d1 - K * np.exp(-r * T) * N_d2
call_price
```
