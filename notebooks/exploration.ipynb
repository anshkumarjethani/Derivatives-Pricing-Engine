```python
S = 100 
K = 100
r = 0.05
sigma = 0.2
T = 1
```

```python
import numpy as np
```

```python
d1 = (np.log(S/K) + (r + sigma**2 / 2)) * T / (sigma * np.sqrt(T))
d2 = d1 - sigma * np.sqrt(T)

d1, d2
```

```python
N_d1 = norm.cdf(d1)
N_d2 = norm.cdf(d2)

N_d1, N_d2
```

```python
call_price = S * N_d1 - K * np.exp(-r * T) * N_d2
call_price
```

```python
from market_data import get_spot_price, filter_liquid_options
from datetime import datetime
import yfinance as yf
```

```python
S = get_spot_price("AAPL")
aapl = yf.Ticker("AAPL")

for expiry in aapl.options:
    days_out = (datetime.strptime(expiry, '%Y-%m-%d') - datetime.now()).days
    if days_out < 20:
        continue
    chain = aapl.option_chain(expiry)
    liquid = filter_liquid_options(chain.calls, min_volume=1, min_open_interest=1)
    if len(liquid) == 0:
        print(expiry, "no liquid calls at all")
        continue
    liquid = liquid.copy()
    liquid['dist'] = abs(liquid['strike'] - S)
    closest = liquid['dist'].min()
    print(expiry, f"closest liquid strike is {closest:.1f} away (limit is {S*0.15:.1f})")
```
